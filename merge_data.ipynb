```python
import pandas as pd
import numpy as np

# 1. Load the exported CSV dataset
df_robots = pd.read_csv('data/RMBR4-2_export_test.csv')
# 2. Create the Shops Dimension Table (PK: Shop_Id)
shops_data = {
    'Shop_Id': [1, 2, 3, 4, 5],
    'Shop Name': ['chassy', 'paint', 'assembly', 'weld', 'plastics'],
    'Plant': ['Plant 1', 'Plant 1', 'Plant 1', 'Plant 1', 'Plant 1'],
    'TKT': [180, 180, 180, 180, 180]  # Takt time (180 seconds)
}
df_shops = pd.DataFrame(shops_data)

# Define valid Rob_Id values to randomly select from
robot_ids = [101, 102, 103, 104, 105]

# 3. Generate random Shop_Ids and Rob_Ids for every row
num_rows = len(df_robots)
random_shop_ids = np.random.choice(df_shops['Shop_Id'].values, size=num_rows)
random_rob_ids = np.random.choice(robot_ids, size=num_rows)

# 4. Insert the random columns at the front
df_robots.insert(0, 'Rob_Id', random_rob_ids)
df_robots.insert(0, 'Shop_Id', random_shop_ids)

# 5. Save to CSV
df_robots.to_csv('data/RMBR4-2_export_test_linked.csv', index=False)
df_shops.to_csv('data/shops.csv', index=False)

# Preview results
print(df_robots.head(10))
```
